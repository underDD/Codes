```python
import pandas as pd
import matplotlib.pyplot as plt
from pypalettes import load_cmap
from pypalettes import get_hex

# Step 1: Read the data from a text file, skipping the first line
file_path = 'tcurve.txt'
data = pd.read_csv(file_path, skiprows=1, delim_whitespace=True, header=None)

# Step 2: Extract the three columns
x = data[0]  # First column
y1 = data[1] # Second column
y2 = data[2] # Third column

# Step 3: Plot the second column vs. the first column
plt.figure(figsize=(10, 5))
plt.plot(x, y2, label='$V_o$ (t)', linewidth=2, color='#388F30FF')
plt.plot(x, y1, label='$V_c$ (t)', linewidth=2, color='#950404FF')
plt.xlabel('Time (s)', fontsize=12)
plt.ylabel('V (V)', fontsize=12)
plt.legend(fontsize=11, frameon=False)


# Customize the axes
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(1)
ax.spines['bottom'].set_linewidth(1)
ax.xaxis.set_tick_params(width=1)
ax.yaxis.set_tick_params(width=1)

plt.savefig('Respuesta1kHz.pdf', format='pdf')

plt.show()
```
